```python
import cv2
```

```python
cv2.__version__
```

```python
front_face_detector = cv2.CascadeClassifier('haarcascade_frontalface_alt.xml')
```

### Single Object Detection

```python
image = cv2.imread('surya.jpg')
image_grey = cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
bounding_box = front_face_detector.detectMultiScale(image_grey)
for x,y,w,h in bounding_box:
    cv2.rectangle(image_grey,(x,y),(x+w,y+h),(255,0,0),3)
cv2.imshow("surya",image_grey)
cv2.waitKey()
```

### Multiple Object Detection

```python
image = cv2.imread('multiple_face.jpg')
image_grey = cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
bounding_box = front_face_detector.detectMultiScale(image_grey)
for x,y,w,h in bounding_box:
    cv2.rectangle(image_grey,(x,y),(x+w,y+h),(255,0,0),3)
cv2.imshow("surya'S Group",image_grey)
cv2.waitKey()
```

### WEB/CCTV Cam

```python
import cv2
front_face_detector = cv2.CascadeClassifier('haarcascade_frontalface_alt.xml')
video = cv2.VideoCapture(0)
success,frame = video.read()
while True:
    # if success == True:
        success,frame = video.read()
        image_grey = cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
        bounding_box = front_face_detector.detectMultiScale(image_grey)
        for x,y,w,h in bounding_box:
            cv2.rectangle(image_grey,(x,y),(x+w,y+h),(255,255,0),3)
        cv2.imshow("surya'S Group",image_grey)
        key = cv2.waitKey(5)
        if key == 113 or key==81:
            break
video.release()
cv2.destroyAllWindows()
```
